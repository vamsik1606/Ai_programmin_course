### task 5 template matching

```python
from skimage import data
from skimage.feature import match_template

image_coins_task5 = data.coins()
coin = image_coins_task5[170:220, 75:130]  # Extract a template of a coin from the image
```

```python
io.imshow(image_coins_task5)
```

```python
io.imshow(coin)
```

```python
result = match_template(image_coins_task5, coin)
ij = np.unravel_index(np.argmax(result), result.shape)
x, y = ij[::-1]
#what happened here is that we are using the match_template 
# function from the skimage.feature module to find the location of a 
# template (in this case, a coin) in an image (in this case, the coins image). 
# The match_template function computes the normalized cross-correlation between the 
# template and the image, and returns a 2D array of correlation coefficients. 
# The np.argmax function is then used to find the index of the maximum correlation 
# coefficient in the result array, which corresponds to the location of the best match 
# for the template in the image. Finally, np.unravel_index is used to convert the index 
# of the maximum correlation coefficient into 2D coordinates (x, y) in the original image.
```

```python
fig = plt.figure(figsize=(8, 3))
ax1 = plt.subplot(1, 3, 1)
ax2 = plt.subplot(1, 3, 2)
ax3 = plt.subplot(1, 3, 3,sharex=ax2, sharey=ax2)
ax1.imshow(coin, cmap=plt.cm.gray)
ax1.set_axis_off()
ax1.set_title('template')
ax2.imshow(image_coins_task5, cmap=plt.cm.gray)
ax2.set_axis_off() 
ax2.set_title('image')
# highlight matched region
hcoin, wcoin = coin.shape
rect = plt.Rectangle((x, y), wcoin, hcoin, edgecolor='r', facecolor='none')
ax2.add_patch(rect)

ax3.imshow(result, cmap=plt.cm.gray)
ax3.set_axis_off()
ax3.set_title('`match_template`\nresult')
# highlight matched region
ax3.autoscale(False)
ax3.plot(x, y, 'o', markeredgecolor='r', markerfacecolor='none', markersize=10)

plt.show()
```

```python
#implement template matching algorithm mathematically
import imageio

import numpy as np

image_task6 = image_coins_task5.astype(float)
template_task6 = image_task6[170:220, 75:130].astype(float)  

H, W = image_task6.shape
h, w = template_task6.shape

# Create the result matrix to store matching scores
result = np.zeros((H - h + 1, W - w + 1))

# Subtract the template's mean intensity
T = template_task6 - np.mean(template_task6)

# Calculate the template's norm once
T_norm = np.sqrt(np.sum(T ** 2))

# Slide the template over the image
for y in range(H - h + 1):
    for x in range(W - w + 1):

        # Extract the image patch at the current position
        patch = image_task6[y:y+h, x:x+w]

        # Subtract the patch's mean intensity
        P = patch - np.mean(patch)

        # Calculate the denominator
        denominator = T_norm * np.sqrt(np.sum(P ** 2))

        # Calculate normalized cross-correlation
        if denominator > 0:
            result[y, x] = np.sum(T * P) / denominator
        else:
            result[y, x] = 0

# Find the position with the highest score
ij = np.unravel_index(np.argmax(result), result.shape)
y, x = ij

print("Best match (x, y):", x, y)
print("Best matching score:", result[y, x])
```
